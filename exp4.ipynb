# Experiment: Statistical Estimation and Hypothesis Testing
# Dataset: Pima Indians Diabetes Dataset

```python
import pandas as pd
import numpy as np
from scipy import stats
```

# 1. Load Dataset

```python

df = pd.read_csv(r"/content/diabetes.csv")

print("First 5 records:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)
```

 2. Point Estimation

```python

# Sample Mean of Glucose
glucose_mean = df["Glucose"].mean()

print("\nSample Mean of Glucose:")
print(glucose_mean)


# Sample Proportion of diabetic patients
diabetic_count = df["Outcome"].sum()
total_count = len(df)

sample_proportion = diabetic_count / total_count

print("\nNumber of diabetic patients:", diabetic_count)
print("Total patients:", total_count)
print("Sample Proportion:", sample_proportion)
print("Sample Proportion (%):", sample_proportion * 100)
```


# 3. 95% Confidence Interval for Mean Glucose




```python

glucose = df["Glucose"]

mean = glucose.mean()
std = glucose.std()
n = len(glucose)

confidence = 0.95

# Standard Error
se = std / np.sqrt(n)

# t critical value
t_critical = stats.t.ppf((1 + confidence) / 2, df=n - 1)

# Confidence interval
lower = mean - t_critical * se
upper = mean + t_critical * se

print("\n95% Confidence Interval for Mean Glucose:")
print("Lower Limit:", lower)
print("Upper Limit:", upper)
```

# 4. One-Sample t-Test

```python
# Test whether mean glucose differs from 120

hypothesized_mean = 120

t_stat, p_value = stats.ttest_1samp(
    df["Glucose"],
    hypothesized_mean
)

print("\nOne-Sample t-Test")
print("Test Statistic:", t_stat)
print("p-value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("Decision: Reject H0")
    print("Conclusion: Mean glucose is significantly different from 120.")
else:
    print("Decision: Fail to Reject H0")
    print("Conclusion: There is insufficient evidence that mean glucose differs from 120.")
```

# 5. Two-Sample t-Test

```python
# Compare BMI of diabetic and non-diabetic patients

bmi_non_diabetic = df[df["Outcome"] == 0]["BMI"]
bmi_diabetic = df[df["Outcome"] == 1]["BMI"]

print("\nMean BMI - Non-Diabetic:", bmi_non_diabetic.mean())
print("Mean BMI - Diabetic:", bmi_diabetic.mean())

t_stat, p_value = stats.ttest_ind(
    bmi_non_diabetic,
    bmi_diabetic,
    equal_var=False
)

print("\nTwo-Sample t-Test")
print("Test Statistic:", t_stat)
print("p-value:", p_value)

if p_value < alpha:
    print("Decision: Reject H0")
    print("Conclusion: Mean BMI is significantly different between the two groups.")
else:
    print("Decision: Fail to Reject H0")
    print("Conclusion: No significant difference in mean BMI.")
```

# 6. Chi-Square Test

```python
# Create BMI categories

def bmi_category(bmi):
    if bmi < 25:
        return "Normal"
    elif bmi < 30:
        return "Overweight"
    else:
        return "Obese"


df["BMI_Category"] = df["BMI"].apply(bmi_category)

# Create contingency table
contingency_table = pd.crosstab(
    df["BMI_Category"],
    df["Outcome"]
)

print("\nContingency Table:")
print(contingency_table)

# Chi-square test
chi2, p_value, dof, expected = stats.chi2_contingency(
    contingency_table
)

print("\nChi-Square Test")
print("Chi-Square Statistic:", chi2)
print("Degrees of Freedom:", dof)
print("p-value:", p_value)

if p_value < alpha:
    print("Decision: Reject H0")
    print("Conclusion: BMI category and diabetes outcome are associated.")
else:
    print("Decision: Fail to Reject H0")
    print("Conclusion: No significant association between BMI category and diabetes outcome.")
```


 # 7. Final Summary



```python
print("\n-----------------------------------------")
print("FINAL SUMMARY")
print("-----------------------------------------")

print("Mean Glucose:", round(glucose_mean, 2))
print("Diabetes Proportion:", round(sample_proportion, 4))
print("95% CI for Glucose:",
      round(lower, 2), "to", round(upper, 2))

print("\nStatistical tests completed:")
print("1. One-Sample t-Test")
print("2. Two-Sample t-Test")
print("3. Chi-Square Test")
```
