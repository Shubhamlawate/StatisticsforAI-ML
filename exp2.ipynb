```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load CSV file
df = pd.read_csv(r"/content/sample_data/diabetes (1).csv")

# Display first 5 rows
print(df.head())
```

```python
# Shape of dataset
print("Shape:", df.shape)

# Column names
print("\nColumns:")
print(df.columns)

# Data types
print("\nData Types:")
print(df.dtypes)

# General information
print("\nDataset Info:")
print(df.info())
```

```python
print("Mean")
print(df.mean(numeric_only=True))
```

```python
print("\nMedian")
print(df.median(numeric_only=True))
```

```python
print("\nMode")
print(df.mode().iloc[0])
```

```python
print("\nMinimum")
print(df.min(numeric_only=True))
```

```python
print("\nMaximum")
print(df.max(numeric_only=True))
```

```python
print("\nVariance")
print(df.var(numeric_only=True))
```

```python
print("\nStandard Deviation")
print(df.std(numeric_only=True))
```

```python
print("\nSummary Statistics")
print(df.describe())
```

```python
df.hist(figsize=(14,10), bins=20, edgecolor='black')
plt.suptitle("Histogram of Numerical Variables")
plt.show()
```

```python
plt.figure(figsize=(14,8))
sns.boxplot(data=df)
plt.title("Boxplot of Numerical Variables")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.countplot(x='Outcome', data=df)

plt.title("Frequency of Diabetes Outcome")
plt.xlabel("Outcome")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.scatterplot(
    x='Glucose',
    y='BMI',
    hue='Outcome',
    data=df
)

plt.title("Glucose vs BMI")
plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.scatterplot(
    x='Age',
    y='Glucose',
    hue='Outcome',
    data=df
)

plt.title("Age vs Glucose")
plt.show()
```

```python
selected_columns = ['Glucose','BMI','Age','Insulin','Outcome']

sns.pairplot(
    df[selected_columns],
    hue='Outcome'
)

plt.show()
```
